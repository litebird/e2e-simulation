```python
save_folder = "../ancillary/"
```

```python
def get_det_names(instrument, channel, wafers, det_numbers):
    # get channel info
    channelinfo = lbs.FreqChannelInfo.from_imo(
        url="/releases/v1.3/satellite/" + instrument + "/" + channel + "/channel_info",
        imo=imo,
    )

    # print(channelinfo.net_detector_ukrts)
    # print(channelinfo.net_channel_ukrts)

    # get instrument number
    if instrument == "LFT":
        instr_number = "000"
    if instrument == "MFT":
        instr_number = "001"
    if instrument == "HFT":
        instr_number = "002"

    # empty list for output
    det_names_output = []

    # if you want all the detectors in the channel (so wafers==None)...
    if wafers is None:
        for det_name in channelinfo.detector_names:
            # if(det_name[-1]=='T'):
            det_names_output.append([instrument, channel, None, None, None, det_name])

    # else...
    else:
        # loop over selected wafers
        for iwafer, wafer in enumerate(wafers):
            # loop over selected det numbers for the wafer
            for det_number in det_numbers[iwafer]:
                # create beginning of det name (like xxx_xxx_xxx)
                det_str = (
                    instr_number
                    + "_"
                    + str(wafer).zfill(3)
                    + "_"
                    + str(det_number).zfill(3)
                )
                # loop over real det names
                for det_name in channelinfo.detector_names:
                    # check beginning of det name and select only top detectors
                    if det_str == det_name[:11]:  # and det_name[-1]=='T'):
                        # append info that we want to save
                        # det_names_output.append([instrument,channel,det_name])
                        det_names_output.append(
                            [instrument, channel, None, None, None, det_name]
                        )

    # get noise factor, rescaled for 1 year and "number of selected detectors/number of total detectors"
    ### !!! Number_det and Scaled_NET consider one year, T and B detectors !!! ###
    imo_net = channelinfo.net_detector_ukrts
    rescale_factor = np.sqrt(
        1 / 3 * len(det_names_output) / len(channelinfo.detector_names)
    )
    for det_name_output in det_names_output:
        det_name_output[2] = str(imo_net)
        det_name_output[3] = (
            str(len(det_names_output)) + "/" + str(len(channelinfo.detector_names))
        )
        det_name_output[4] = str(np.around(imo_net * rescale_factor, 2))

    # save det names for channel
    det_names_save = np.array(det_names_output)
    np.savetxt(
        save_folder + "detectors_" + instrument + "_" + channel + "_T+B.txt",
        det_names_save,
        delimiter="\t\t",
        header="Telescope     Channel         IMO_NET         Number_det      Scaled_NET      Detector_name",
        fmt="%s",
    )
```
